```python
#Combine everything in a neat Summary
from Bio import SeqIO

# Load the first record
record = next(SeqIO.parse("gene.fna", "fasta"))

#Calculating the gc content
def gc_content(seq):
    g = seq.count("G")
    c = seq.count("C")
    return (g + c) / len(seq) * 100

def reverse_complement(seq):
    return seq.reverse_complement()

def transcribe(seq):
    return seq.transcribe()

def translate(seq):
    return seq.translate()

# Run everything and store results
gc = gc_content(record.seq)
rev_comp = reverse_complement(record.seq)
rna = transcribe(record.seq)
protein = translate(record.seq)

# Print a clean summary report
print("=" * 50)
print(f"GENE SEQUENCE REPORT")
print("=" * 50)
print(f"ID: {record.id}")
print(f"Length: {len(record.seq)} bases")
print(f"GC Content: {gc:.2f}%")
print("-" * 50)
print(f"Original DNA (first 60):     {record.seq[:60]}")
print(f"Reverse Complement (60):     {rev_comp[:60]}")
print(f"RNA (first 60):              {rna[:60]}")
print(f"Protein (first 60):          {protein[:60]}")
print("=" * 50)
```
